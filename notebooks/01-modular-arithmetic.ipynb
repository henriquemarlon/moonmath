# ZK01x — Modular arithmetic

Original examples to begin studying the **MoonMath Manual**. Run this notebook with the **SageMath 10.9** kernel.

Before running each cell, predict its result on paper. `Zmod(n)` represents integers modulo `n`; `ZZ` represents the integers.


```sage
R = Zmod(17)
a, b = R(5), R(14)
assert a + b == R(2)
assert a * b == R(2)
print("5 + 14 mod 17 =", a + b)
print("5 × 14 mod 17 =", a * b)
```

## Multiplicative inverses

The inverse of 5 modulo 17 is 7, because 5 × 7 leaves remainder 1 when divided by 17. An integer `a` has an inverse modulo `n` if and only if `gcd(a, n) = 1`.


```sage
inverse = a**(-1)
assert inverse == R(7)
assert a * inverse == R(1)
print("Inverse of 5 mod 17 =", inverse)
print("Verification:", a * inverse)
```

```sage
g, s, t = xgcd(ZZ(5), ZZ(17))
assert g == 1
assert s*5 + t*17 == g
assert R(s) == inverse
print("Bézout: gcd =", g, "; coefficients =", (s, t))
```

## A composite modulus

Modulo 15, some nonzero residues have no inverse: 3 × 5 = 0 even though both factors are nonzero. Compare this with the prime modulus 17.


```sage
S = Zmod(15)
assert S(3) != 0 and S(5) != 0
assert S(3)*S(5) == 0
units = [k for k in range(15) if gcd(k, 15) == 1]
assert units == [1, 2, 4, 7, 8, 11, 13, 14]
print("Invertible residues mod 15:", units)
```
